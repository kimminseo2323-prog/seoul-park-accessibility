```python
# ==================================================================================
# 8. Landsat NDVI (Plotly 시각화)
# ==================================================================================
print("\n[8] Landsat NDVI 위성영상 데이터")
print("=" * 100)

try:
    import rasterio
    import numpy as np

    ndvi_path = os.path.join(base_path, 'Landsat_위성영상의_정규식생지수_2024', 'L9_NDVI.tif')

    with rasterio.open(ndvi_path) as src:
        print(f"파일명: L9_NDVI.tif")
        print(f"크기 (행x열): {src.height} x {src.width}")
        print(f"밴드 수: {src.count}")
        print(f"좌표계(CRS): {src.crs}")
        print(f"경계(Bounds): {src.bounds}")
        print(f"해상도: {src.res}")
        print(f"\nNDVI 값 범위:")
        ndvi_data = src.read(1)
        print(f"  최소값: {ndvi_data.min():.4f}")
        print(f"  최대값: {ndvi_data.max():.4f}")
        print(f"  평균값: {ndvi_data.mean():.4f}")
        print(f"  표준편차: {ndvi_data.std():.4f}")

        # Plotly로 NDVI 시각화
        print(f"\n📊 인터랙티브 NDVI 시각화 (Plotly):")

        try:
            import plotly.graph_objects as go

            # NDVI 데이터 샘플링 (너무 크면 다운샘플링)
            max_size = 2000
            if ndvi_data.shape[0] > max_size or ndvi_data.shape[1] > max_size:
                step_h = max(1, ndvi_data.shape[0] // max_size)
                step_w = max(1, ndvi_data.shape[1] // max_size)
                ndvi_display = ndvi_data[::step_h, ::step_w]
                print(f"  💡 데이터 크기가 커서 {step_h}x{step_w} 간격으로 샘플링")
            else:
                ndvi_display = ndvi_data

            # 1. NDVI Heatmap
            print(f"\n  [1] NDVI 공간 분포 지도")
            fig = go.Figure(data=go.Heatmap(
                z=ndvi_display,
                colorscale='RdYlGn',
                colorbar=dict(
                    title=dict(
                        text="NDVI",
                        side="right"
                    ),
                    tickmode="linear",
                    tick0=-1,
                    dtick=0.2
                ),
                zmin=-1,
                zmax=1,
                hovertemplate='행: %{y}<br>열: %{x}<br>NDVI: %{z:.4f}<extra></extra>'
            ))

            fig.update_layout(
                title='Landsat 9 NDVI (2024) - 서울시',
                xaxis_title='X (Pixel)',
                yaxis_title='Y (Pixel)',
                width=900,
                height=700,
                yaxis=dict(autorange='reversed')
            )

            fig.show()

            # 2. NDVI 분포 히스토그램
            print(f"\n  [2] NDVI 값 분포 히스토그램")
            fig_hist = go.Figure(data=[go.Histogram(
                x=ndvi_data.flatten(),
                nbinsx=50,
                marker_color='green',
                opacity=0.7,
                name='NDVI 분포'
            )])

            fig_hist.update_layout(
                title='NDVI 값 분포',
                xaxis_title='NDVI',
                yaxis_title='빈도 (픽셀 수)',
                width=800,
                height=400,
                showlegend=True
            )

            fig_hist.show()

            # 3. NDVI 통계 Box Plot
            print(f"\n  [3] NDVI 통계 분석")
            fig_box = go.Figure()
            fig_box.add_trace(go.Box(
                y=ndvi_data.flatten(),
                name='NDVI',
                marker_color='green',
                boxmean='sd'
            ))

            fig_box.update_layout(
                title='NDVI 통계 (Box Plot)',
                yaxis_title='NDVI',
                width=600,
                height=500
            )

            fig_box.show()

        except ImportError:
            print("  ⚠ plotly 패키지가 설치되지 않았습니다.")
            print("  설치: pip install plotly")
            print("  또는: conda install -c conda-forge plotly")

except ImportError:
    print("⚠ rasterio 패키지가 설치되지 않았습니다.")
    print("설치: conda install -c conda-forge rasterio")
    print("\n파일 정보만 출력:")
    ndvi_folder = os.path.join(base_path, 'Landsat_위성영상의_정규식생지수_2024')
    files = os.listdir(ndvi_folder)
    print(f"폴더 내 파일 목록:")
    for f in files:
        file_path = os.path.join(ndvi_folder, f)
        if os.path.isfile(file_path):
            size_mb = os.path.getsize(file_path) / (1024 * 1024)
            print(f"  - {f} ({size_mb:.2f} MB)")
```
